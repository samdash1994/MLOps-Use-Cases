In Python, we don’t have a built-in vector type. So we can represent vectors using either lists or tuples. Even so, there are certain limitations with this kind of representation, but let's explore these anyway.

Let’s start by representing the following vectors using lists

$ \mathbf {v}_1 = 3\hat{i} + 2\hat{j} + 5\hat{k}$

$ \mathbf {v}_2 = 4\hat{i} + 5\hat{j} + 6\hat{k}$

```python
v1 = [3, 2, 5]
v2 = [4, 5, 6]
```

Now, let's try to add vectors

```python
print(v1 + v2)  # Is this vector addition?
```

This is not vector addition, this is list concatenation. Python doesn't know you meant to add element-wise, it just combines the lists.

Let's try the same with tuples



```python
v1 = (3, 2, 5)
v2 = (4, 5, 6)

print(v1 + v2)
```

Again, this is not vector addition

Let’s perform proper element-wise addition of two vectors (represented as tuples)

$\mathbf{v}_1 + \mathbf{v}_2 = (3\hat{i} + 2\hat{j} + 5\hat{k}) + (4\hat{i} + 5\hat{j} + 6\hat{k})$

$\mathbf{v}_1 + \mathbf{v}_2 = (3+4)\hat{i} + (2+5)\hat{j} + (5+6)\hat{k}$

$\mathbf{v}_1 + \mathbf{v}_2 = 7\hat{i} + 7\hat{j} + 11\hat{k}$

```python
def add_vectors(v1, v2):
    if len(v1) != len(v2):
        raise ValueError("Vectors must be of same dimension")
    return tuple(v1[i] + v2[i] for i in range(len(v1)))

print(add_vectors(v1, v2))
```

What if we used lists instead?

```python
v1 = [3, 2, 5]
v2 = [4, 5, 6]

print(add_vectors(v1, v2))  # Also works, but less secure
```

But lists allow mutation, which can change the vector

```python
v1.append(100)
v1
```

This is no longer the same vector. With tuples, this would raise an error. So, tuples enforce vector structure.

Now, let’s perform proper element-wise subtraction of two vectors using tuples

$\mathbf{v}_1 - \mathbf{v}_2 = (3\hat{i} + 2\hat{j} + 5\hat{k}) - (4\hat{i} + 5\hat{j} + 6\hat{k})$

$ \mathbf{v}_1 - \mathbf{v}_2 = (3-4)\hat{i} + (2-5)\hat{j} + (5-6)\hat{k}$

$ \mathbf{v}_1 - \mathbf{v}_2 = -1\hat{i} - 3\hat{j} - 1\hat{k}$

```python
v1 = (3, 2, 5)
v2 = (4, 5, 6)

def subtract_vectors(v1, v2):
    return tuple(v1[i] - v2[i] for i in range(len(v1)))

print(subtract_vectors(v1, v2))
```

Let's see how can we perform dot product of two vectors using loops

$ \mathbf{v}_1 \cdot \mathbf{v}_2 = (3\hat{i} + 2\hat{j} + 5\hat{k}) \cdot (4\hat{i} + 5\hat{j} + 6\hat{k})$

$ \mathbf{v}_1 \cdot \mathbf{v}_2 = (3 \times 4) + (2 \times 5) + (5 \times 6)$

$ \mathbf{v}_1 \cdot \mathbf{v}_2 = 12 + 10 + 30$

$ \mathbf{v}_1 \cdot \mathbf{v}_2 = 52$

```python
dot_product = 0

for i in range(len(v1)):
    dot_product += v1[i] * v2[i]

print("Dot Product:", dot_product)
```

Let's calculate the magnitude of vector $\mathbf{v}_1$

$|| \mathbf{v}_1||  = \sqrt{3^2 + 2^2 + 5^2} = \sqrt{9 + 4 + 25} = \sqrt{38} \approx 6.16$

```python
def magnitude(vector):
    sum_of_squares = 0
    for component in vector:
        sum_of_squares += component**2
    magnitude = sum_of_squares**0.5
    return magnitude

round(magnitude(v1), 2)
```
